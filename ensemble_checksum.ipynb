```python
from utils_ensemble import load_ensemble_tuples
import os
import numpy as np
import glob
```

```python
tuples = load_ensemble_tuples()

base_path = "/gpfs/wolf2/cades/cli185/proj-shared/zdr/ERW/output/UQ/pft1"

bad_ens = []
bad_folders = []
for ens in range(1, 2001):
    ens_str = f"{ens:04d}"
    pattern = os.path.join(base_path, f"*_conus_ICB20TRCNPRDCTCBC_erw_pft1_ens{ens_str}")
    dirs = glob.glob(pattern + "/")
    if not dirs:
        continue
    folder = sorted(dirs)[-1]
    file_pattern = os.path.join(folder, "run", "*.h1.*.nc")
    files = glob.glob(file_pattern)
    if len(files) < 57:
        bad_folders.append(folder)
        bad_ens.append(ens)
```
