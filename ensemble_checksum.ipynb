```python
from utils_ensemble import load_ensemble_tuples
import os
import numpy as np
import glob
```

```python
tuples = load_ensemble_tuples()

base_path = "/gpfs/wolf2/cades/cli185/proj-shared/zdr/ERW/output/UQ/pft1"

bad_ens = []
# Total should go from 0 to 2275 (2276 runs)
#for ens in range(0 2276):
for ens in range(0, 2276):
    ens_str = f"{ens:04d}"
    pattern = os.path.join(base_path, f"*_conus_ICB20TRCNPRDCTCBC_erw_pft1_ens{ens_str}")
    dirs = glob.glob(pattern + "/")
    if not dirs:
        bad_ens.append(ens)
        continue
    folder = sorted(dirs)[-1]
    file_pattern = os.path.join(folder, "run", "*.h1.*.nc")
    files = glob.glob(file_pattern)
    if len(files) < 57:
        bad_ens.append(ens)
```

```python
###############################################################
# Filter out the ensemble members before 20250530
###############################################################

# Those are inconsistent
bad_ens = []
# Total should go from 0 to 2275 (2276 runs)
#for ens in range(0 2276):
for ens in range(0, 2276):
    ens_str = f"{ens:04d}"
    pattern = os.path.join(base_path, f"*_conus_ICB20TRCNPRDCTCBC_erw_pft1_ens{ens_str}")
    dirs = glob.glob(pattern + "/")
    if not dirs:
        bad_ens.append(ens)
        continue
    folder = sorted(dirs)[-1]
    prefix = int(folder.split('/')[-2][:8])

    if prefix < 20250530:
        bad_ens.append(ens)
```

```python
tuples.loc[bad_ens, 'inconsistent'] = True
tuples.index.name = 'ensemble_id'
tuples.to_csv(os.path.join(os.environ['PROJDIR'], 'ERW_LDRD', 'results', 
                           'erw_ensemble_record.csv'))
```
